## 4. Feature & Target Variable Classification
We programmatically categorize the columns into Target ($Y$), Numerical Features, Categorical Features, and Identifiers.

```python
# Define target variable
target_col = 'Delivery_Time_min'

# Define metadata / identifier column
id_col = 'Order_ID'

# Identify numerical feature columns
numerical_features = [
    col for col in df.select_dtypes(include=[np.number]).columns
    if col not in [target_col, id_col]
]

# Identify categorical feature columns
categorical_features = df.select_dtypes(include=['object', 'str']).columns.tolist()

print(f"Target Variable (Y):\n  -> {target_col}\n")
print(f"Identifier (Excluded from features):\n  -> {id_col}\n")
print(f"Numerical Features ({len(numerical_features)}):\n  -> {numerical_features}\n")
print(f"Categorical Features ({len(categorical_features)}):\n  -> {categorical_features}")
```

### Analysis & Observations
* **Supervised Learning Task**: This is a **Continuous Regression** problem because the target `Delivery_Time_min` is numeric and continuous.
* **Predictor Set ($X$)**:
  * 3 Numerical inputs: `Distance_km`, `Preparation_Time_min`, `Courier_Experience_yrs`
  * 4 Categorical inputs: `Weather`, `Traffic_Level`, `Time_of_Day`, `Vehicle_Type`
* **Exclusion of `Order_ID`**: `Order_ID` must strictly be excluded from model training to prevent target leakage and spurious index correlations.

## 8. Range Validation & Sanity Checks
We verify that numerical values do not violate physical boundaries or business logic.

```python
# Range and sanity checks
sanity_checks = {
    'Negative Distance': (df['Distance_km'] < 0).sum(),
    'Zero or Negative Prep Time': (df['Preparation_Time_min'] <= 0).sum(),
    'Negative Courier Experience': (df['Courier_Experience_yrs'] < 0).sum(),
    'Negative Delivery Time': (df['Delivery_Time_min'] <= 0).sum()
}

for check, violations in sanity_checks.items():
    status = 'PASSED (No violations)' if violations == 0 else f'FAILED ({violations} violations)'
    print(f"{check:30s}: {status}")
```

### Analysis & Observations
* **Sanity Checks Passed**:
  * No negative distances or zero prep times.
  * No negative courier experience values.
  * No negative or zero delivery times.
* The data conforms to physical delivery constraints, and no corrupt or nonsensical values were detected.
